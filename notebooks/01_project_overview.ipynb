# BrainHack Python fNIRS MA Analysis Pipeline

## Project overview

This project aims to build a Python-based fNIRS analysis pipeline for the BrainHack final project.

The research topic focuses on comparing brain activation during morphological awareness (MA) tasks between typically developing children in lower grades (G1–3) and upper grades (G4–6).

This repository is separate from the original MATLAB-based thesis pipeline. The goal of this project is to reconstruct and demonstrate a reproducible Python-based workflow for fNIRS data analysis.


## Research question

The main research question is:

Do typically developing children in lower grades (G1–3) and upper grades (G4–6) show different brain activation patterns during morphological awareness tasks?

More specifically, this project will examine:

1. Whether MA-related brain activation differs between lower-grade and upper-grade children.
2. Which fNIRS channels show significant group differences.
3. Whether the results remain significant after multiple-comparison correction.
4. Whether MA-related activation features can be used in an exploratory machine learning model to classify grade groups.


## Data privacy and ethics

This project uses lab-based fNIRS data from children. Therefore, raw data and subject information will not be uploaded to GitHub.

The following files will be excluded from version control:

- raw fNIRS data
- subject-level demographic information
- Excel or CSV files containing participant information
- MATLAB data files
- any file containing identifiable or sensitive metadata

The repository will only include code, documentation, reproducible workflow descriptions, and non-sensitive demonstration materials.


## Planned analysis pipeline

The planned Python-based analysis pipeline includes:

1. Organize project structure and GitHub repository.
2. Load or simulate fNIRS channel-level activation data.
3. Define grade groups: lower grades G1–3 and upper grades G4–6.
4. Compare MA-related activation between groups.
5. Generate uncorrected statistical results.
6. Apply multiple-comparison correction, such as FWE correction.
7. Visualize significant channels and brain activation patterns.
8. Conduct an exploratory machine learning analysis using MA-related activation features.
9. Document the workflow for reproducibility.
